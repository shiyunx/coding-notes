**Python dictionary to json file**

```python
import json                              # Import JSON module

data = {                                 # Create Python dictionary
    "username": "iuser",                 # Add username
    "role": "admin"                      # Add user role
}

with open("user.json", "w") as file:     # Open/create JSON file in write mode
    json.dump(data, file)                # Write dictionary into JSON file

print("File created")                    # Confirm file creation
```

**Access nested json**

```python
import json

# Create a JSON string
json_text = '''
{
    "username": "iuser",
    "details": {
        "email": "iuser@example.com",
        "age": 30
    }
}
'''

# Convert JSON string to dictionary
data = json.loads(json_text)

# Access username
print(data["username"])

# Access nested email
print(data["details"]["email"])

# Access nested age
print(data["details"]["age"])
```

**Json list/array**

```python
import json                                      # Import JSON module

# Create JSON string
json_text = '''
{
    "users": [
        {
            "username": "iuser",
            "role": "admin"
        },
        {
            "username": "juser",
            "role": "user"
        }
    ]
}
'''

# Convert JSON to Python dictionary
data = json.loads(json_text)

# Access first user
print(data["users"][0])

# Access first user's username
print(data["users"][0]["username"])

# Access second user's role
print(data["users"][1]["role"])
```

**Loop through json list**

```python
import json                                      # Import JSON module

# Create a JSON string
json_text = '''
{
    "users": [
        {
            "username": "iuser",
            "role": "admin"
        },
        {
            "username": "juser",
            "role": "user"
        }
    ]
}
'''

# Convert JSON string to a Python dictionary
data = json.loads(json_text)

# Loop through each user in the "users" list
for user in data["users"]:

    # Print the username
    print(user["username"])

    # Print the user's role
    print(user["role"])
```

**Get with default value**

```python
import json                                      # Import JSON module

json_text = '{"username": "iuser"}'              # Create JSON string

data = json.loads(json_text)                     # Convert JSON to dictionary

email = data.get("email", "No email")            # Return "No email" if key is missing

print(email)                                     # Print the result
```

**Check if key exists**

```python
import json                                      # Import JSON module

json_text = '{"username": "iuser"}'              # Create JSON string

data = json.loads(json_text)                     # Convert JSON to dictionary

if "email" in data:                              # Check whether email key exists
    print(data["email"])                         # Print email if it exists
else:                                            # Run if email does not exist
    print("Key not found")                       # Print error message
```

**Handle missing key with try/except**

```python
import json                                      # Import JSON module

json_text = '{"username": "iuser"}'              # Create JSON string

data = json.loads(json_text)                     # Convert JSON to dictionary

try:                                             # Start error-handling block
    print(data["email"])                         # Try to access missing key
except KeyError:                                 # Catch missing-key error
    print("Key not found")                       # Print error message
```

**Delete a key**

```python
import json                                      # Import JSON module

data = {                                         # Create Python dictionary
    "username": "iuser",                         # Add username
    "password": "abc123",                        # Add password
    "is_user": True                              # Add user status
}

data.pop("password", None)                       # Remove password key safely

print(data)                                      # Print updated dictionary
```

**Delete key and save json file**

```python
import json                                      # Import JSON module

with open("user.json", "r") as file:             # Open JSON file in read mode
    user = json.load(file)                       # Convert JSON file to dictionary

user.pop("password", None)                       # Remove password key safely

with open("user.json", "w") as file:             # Open JSON file in write mode
    json.dump(user, file, indent=4)              # Save updated dictionary as formatted JSON

print("Password deleted")                        # Confirm the update
```

**Handle invalid json**

```python
import json                                      # Import JSON module

json_text = '{"username": "iuser",}'             # Create invalid JSON

try:                                             # Start error-handling block
    data = json.loads(json_text)                 # Try to convert JSON string
    print(data)                                  # Print converted data

except json.JSONDecodeError:                     # Catch invalid JSON error
    print("Invalid JSON")                        # Print error message
```
