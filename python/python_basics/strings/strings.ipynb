**Strings**

```python
# Print a alphabet
print('Q')
```

```python
# Print text
print("hello world!")
```

**Multiline strings**

```python
paragraph = """This is line 1
This is line 2
This is line 3"""

print(paragraph)
```

**F-strings**

```python
# Print f-strings
first_name = "firstname"
last_name = "lastname"
full_name = f"{first_name} {last_name}"

print(f"My name is {full_name}.")
```

**Strip whitespace**

```python
# Remove text whitespace
text = " whitespace "

print(text.rstrip()) # " whitespace"
print(text.lstrip()) # "whitespace "
print(text.strip())  # "whitespace"
```

**Center, ljust & rjust**

```python
text = "String formatting"

# Center aligned string
print ("Center aligned string:")
print ( text.center(30) )
print ()

# Left aligned string
print ("Left aligned string:")
print ( text.ljust(30) )
print ()

# Right aligned string
print ("The right aligned string:")
print ( text.rjust(30) )
```

**Uppercase & Lowercase**

```python
# Print uppercase & lowercase
message = "Hello World"

print(message.upper())
print(message.lower())
```

**Title**

```python
# Print title uppercase
message = "introduction"

print(message.title())
```

**Replace**

```python
message = "Hello World"
updated_message = message.replace("World", "Python")

print(updated_message)
```

**Split**

```python
text = "user1,user2,user3"

users = text.split(",")

print(users)
```

**Join**

```python
fruits = ["Apple", "Banana", "Mango", "Pineapple", "Orange"]

print(", ".join(fruits))
```

**Length**

```python
# Print string length
country = "Singapore"

print(len(country))
```

```python
# Print string list
fruits = ['apple', 'berries', 'durian']

print(len(fruits))
```

**Index**

```python
# Print string index
language = "Python"

print(language[3])
```

```python
# Print string index
text = "Python"

print(text[0])
print(text[1])
print(text[-1])
```

```python
# Insert 'durian' at index 1
fruits = ['apple', 'berries']
fruits.insert(1, 'durian')

print(fruits)
```

**Slice**

```python
text = "Python"

print(text[0:3])
```

```python
# Slice list index 1 and index 2
# index = [1, 2, 3]
fruits = ['apple', 'berries', 'durian']

print(fruits[1:3])
```

```python
# Slice string
# string[start:stop:step]

text = "Exact characters using string slice."

print(text[0:3])
```

**Concatenate**

```python
first = "firstname"
second = "lastname"
# Combine first and second strings with a space in between
full_name = first + " " + second

print(full_name)
```

**Zfill**

```python
num = "1"
print(num.zfill(3))
```
