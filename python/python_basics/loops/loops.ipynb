**For loop through items**

```python
# Create a list of colours
colours = ["red", "orange", "blue", "green", "white"]

# Loop through each colour in the list
for colour in colours:
    # Print the current colour
    print(colour)
```

```python
# Store the string "Python" in the variable 'word'
word = "Python"

# Loop through each character in the string
for letter in word:
    print(letter)
```

**For loop: range**

```python
for number in range(1, 5):
    # start = 1 → begin at 1
    # stop = 5 → stop before 5
    print(number)
```

```python
for number in range(0, 5, 1):
    # number starts at 0, stops before 5, and increases by 1
    print(number)
```

```python
# countdown with range
for number in range(5, 0, -1):
    print(number)
```

**For loop: list indexes**

```python
colours = ["red", "orange", "blue", "green", "white"]

# range(len(colours)) generates the index numbers: 0, 1, 2, 3, 4
for index in range(len(colours)):
    # index = position in the list
    # colours[index] = value at that position
    print(index, colours[index])

    print(range(len(colours)))
```

**For loop: list slice**

```python
colours = ["red", "orange", "blue", "green", "white"]

# Loop through items at index 1 up to (but not including) index 3
for colour in colours[1:3]:
    print(colour)
```

**While loop**

```python
# Print numbers from 1 to 10
count = 1

# Continue the loop while count is 10 or less
while count <= 10:
    print(count)
    count += 1  # Increase count by 1
```

**While loop: count backwards**

```python
count = 5

# Continue the loop while count is 1 or greater
while count >= 1:
    print(count)
    count -= 1    # Decrease count by 1
```

**While loop: count downtimer**

```python
# Countdown from 10 to 1
count = 10

# Continue the loop while count is greater than 0
while count > 0:
    print(count)
    count -= 1  # Decrease count by 1

print("Stop")  # Print "Stop" when the countdown reaches 0
```

**While loop: even numbers**

```python
# Start with the first even number
number = 2

# Continue while number is 10 or less
while number <= 10:
    print(number)
    number += 2    # Increase by 2 to get the next even number
```

**While loop: odd numbers**

```python
# Start with the first odd number
number = 1

# Continue while number is 7 or less
while number <= 7:
    print(number)
    number += 2    # Increase by 2 to get the next odd number
```

**While loop: sum**

```python
# Start counting from 1
number = 1

# Start the total at 0
total = 0

# Continue while number is 5 or less
while number <= 5:
    total += number  # Add the current number to total
    number += 1      # Move to the next number

print(total)
```

```python
# Calculate the sum of numbers from 1 to 10

num = 1       # Start counting from 1
total = 0     # Store the running total

# Continue the loop while num is less than or equal to 10
while num <= 10:
    total += num    # Add the current number to the running total
    num += 1        # Increase num by 1

print("Total:", total)
```

**While loop: loop through list**

```python
# Create a list of fruits
fruits =  ["apple", "banana", "orange"]

# Start at the first position in the list
index = 0

# Continue while index is within the list
while index < len(fruits):
    print(fruits[index])  # Print the fruits at the current position
    index += 1            # Move to the next position
```

**Break: stop the loop**

```python
num = 1

# Continue the loop until the break statement is reached
while True:
    # Print the current number
    print(num)

    # Stop the loop when the number reaches 10
    if num == 10:
        break

    # Increase the number by 1
    num += 1
```

**Continue: skip current iteration**

```python
# Skip number 3
num = 0

# Continue the loop while num is less than 10
while num < 10:

    # Increase num by 1
    num += 1

    # Check if num is 3
    if num == 3:
        # Skip the rest of the current loop iteration and go back to the while condition
        continue

    print(num)
```

**Infinite loop**

```python
# Start an infinite loop
# The loop will continue running until "break" is executed
while True:
    print("\nMenu")
    print("1. On")
    print("2. Exit")

    # Prompt the user to choose an option
    choice = input("Choose an option: ")

    # If the user chooses 1, turn the system on
    if choice == "1":
        print("System is On")

    # If the user chooses 2, exit the loop
    elif choice == "2":
        break

# Print after the loop is stopped by "break"
print("Off")
```

**Nested loops**

```python
# Outer loop counter
i = 1

# Outer loop runs while i is less than or equal to 2
while i <= 2:

    # Print the current outer loop value
    print("Outer loop:", i)

    # Reset the inner loop counter to 1
    # This happens every time the outer loop runs
    j = 1

    # Inner loop runs while j is less than or equal to 3
    while j <= 3:

        # Print the current inner loop value
        print("  Inner loop:", j)

        # Increase j by 1 to move to the next inner loop value
        j += 1

    # Increase i by 1 to move to the next outer loop value
    i += 1
```

**While loop and user input**

```python
# Prompt the user to enter a number
number = int(input("Enter a number: "))

# Continue while the number is not 0
while number != 0:
    print("You entered:", number)

    # Prompt for another number
    number = int(input("Enter another number (0 to stop): "))

# Print after the user enters 0
print("Loop ended.")
```
