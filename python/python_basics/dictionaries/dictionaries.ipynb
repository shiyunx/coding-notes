```python
# Nested dictionary with multiple users
# The outer dictionary stores multiple users
users = {
    "user1": {                         # user1 is a key in the outer dictionary
        "username": "admin",
        "email": "admin@mail.com",
        "role": "Administrator",
        "active": True
    },

    "user2": {                         # user2 is another key in the outer dictionary
        "username": "staff",
        "email": "staff@mail.com",
        "role": "Support",
        "active": False
    }
}


# Access a value from the nested dictionary
# First access "user1", then access "role"
print(users["user1"]["role"])

# First access "user2", then access "role"
print(users["user2"]["role"])
```

**Dictionary comprehension**

```python
temperatures = {
    "Monday": 30,
    "Tuesday": 31,
    "Wednesday": 29,
    "Thursday": 32
}

# Dictionary comprehension
fahrenheit = {
    day: (temp * 9/5) + 32
    for day, temp in temperatures.items()
}

print(fahrenheit)
```

**Dictionary comprehension with condition**

```python
temperatures = {
    "Monday": 30,
    "Tuesday": 31,
    "Wednesday": 29,
    "Thursday": 32
}

# Dictionary comprehension
fahrenheit = {
    day: (temp * 9/5) + 32
    for day, temp in temperatures.items()
}

print(fahrenheit)
```

```python
temperatures = {
    "Monday": 30,
    "Tuesday": 31,
    "Wednesday": 29,
    "Thursday": 32
}

# Dictionary comprehension with condition
hot_days = {
    day: temp
    for day, temp in temperatures.items()
    if temp >= 30
}

print(hot_days)
```

```python
# Dictionaries
user1 = {
    "user_id": 1001,
    "username": "user1",
    "is_active": True,
    "login_attempts": 3
}

user2 = {
    "user_id": 1002,
    "username": "user2",
    "is_active": False,
    "login_attempts": 1
}

# Access values
print(user1["username"])
print(user1["is_active"])

print(user2["username"])
print(user2["is_active"])

# Check data types
print(type(user1["username"]))
print(type(user1["is_active"]))
```
