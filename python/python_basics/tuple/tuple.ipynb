Tuple `( )` cannot change (immutable).


**Empty tuple**

```python
# Empty tuple
empty_tuple = ()
```

```python
# Tuple with values
numbers = (1, 2, 3, 4, 5)

print(numbers)
```

**Length**

```python
fruits = ("apple", "banana", "orange")

# Access tuple items
print(fruits[0])
print(fruits[-1])
print(fruits[1:3])
```

```python
# Returns the length of the tuple items
print(len(fruits))
```

**Count**

```python
count_tuple = (1, 2, 3, 3, 5)

# Count how many times a specific value appears
print(count_tuple.count(3))
```

**Index**

```python
t = (10, 20, 30, 40)

# Find the index position of the first occurrence of a value
print(t.index(30))
```
