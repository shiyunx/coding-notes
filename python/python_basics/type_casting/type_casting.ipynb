Type casting means converting a value from one data type to another.

```python
# String
username = "user1"
# Integer
age = 10
# Float
marks = 63.6
# Boolean
student = True

# Check the data type of each variable
print(type(username))
print(type(age))
print(type(marks))
print(type(student))
```

**String to Interger**

```python
x = "10"
y = int(x)

print(y)
print(type(y))
```

**Interger to String**

```python
x = 10
y = str(x)

print(y)
print(type(y))
```

**String to Float**

```python
x = "10.5"
y = float(x)

print(y)
print(type(y))
```

**Interger to Float**

```python
x = 10
y = float(x)

print(y)
```

**Float to Interger**

```python
x = 10.5
y = int(x)

print(y)
```

**List to Tuple**

```python
fruits = ["Apple", "Berries"]
result = tuple(fruits)

print(result)
```

**Tuple to List**

```python
fruits = ("Apple", "Berries")
result = list(fruits)

print(result)
```

**List to Set**

```python
numbers = [1, 2, 2, 3, 3]
result = set(numbers)

print(result)
```

**String to List**

```python
word = "Hello"
result = list(word)

print(result)
```

**List of Tuples to Dictionary**

```python
data = [("username", "Mario"), ("status", "Logged in")]
result = dict(data)

print(result)
```

**Integer to Boolean**

```python
print(bool(1))
print(bool(0))
```

**String to Boolean**

```python
print(bool("Hello"))
print(bool(""))
```
