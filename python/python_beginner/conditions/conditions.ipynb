**if**

```python
password_length = 10

# Check if the password is at least 8 characters long
if password_length >= 8:
    print("Strong password")
```

```python
# Store the user's role
user_role = "admin"

# Check if the user has admin access
if user_role == "admin":
    print("Access granted")
```

```python
# Set the user's role
user_role = "guest"

# Check if the user is not an admin
if user_role != "admin":
    print("Access restricted")
```

```python
# Store the employee's department
department = "HR"

# HR or IT staff can access employee records
if department == "HR" or department == "IT":
    print("Access to employee records granted")
else:
    print("Access denied")
```

**if else**

```python
# Set the user's role
user_role = "guest"

# Check if the user is an admin
if user_role == "admin":
    print("Access granted")
else:
    print("Access denied")
```

```python
# Store the user's department
department = "IT"

# Store the user's role
user_role = "engineer"

# Only IT engineers get access
if department == "IT" and user_role == "engineer":
    # Grant access if both conditions are true
    print("Access to IT systems granted")
else:
    print("Access denied")
```

```python
# Set the user's account status
is_active = True

# Set the user's role
user_role = "staff"

# Only active users can log in
if is_active:
    # Check if the user is a staff member
    if user_role == "staff":
        print("Welcome staff")
    else:
        print("Welcome user")
else:
    # Block inactive users from accessing the system
    print("Account inactive. Access blocked.")
```

```python
# Create a list containing different colour strings
colours = ["red", "orange", "blue", "green", "white"]

# Loop through each colour in the list
for colour in colours:

    # Check if the current colour is "orange"
    if colour == "orange":
        # Convert "orange" to uppercase
        print(colour.upper())
    else:
        # Convert the colour to title case
        print(colour.title())
```

**if elif else**

```python
# Set the user's role
user_role = "user2"

# Check the user's role and assign the appropriate access level
if user_role == "user1":
    print("Full access")
elif user_role == "user2":
    print("Limited access")
elif user_role == "user3":
    print("Basic access")
else:
    print("No access")
```

```python
# Store the current system status
status = "error"

# Check if the status is "success"
if status == "success":
    print("System is working normally")

# If status is not "success", check if it is "warning"
elif status == "warning":
    print("Check the system")

# If status is not "success" or "warning", check if it is "error"
elif status == "error":
    print("Investigate the incident")

else:
    print("Unknown status")
```

**nested if**

```python
# Store the person's age
age = 30

# Store whether the person has a valid ID
has_id = True

# Check if the person is 18 or older
if age >= 18:

    # Check if the person has an ID
    if has_id:
        print("Access allowed")
    else:
        print("ID required")

else:
    print("Access denied")
```

**nested if else**

```python
username = "admin"
password = "1234"

if username == "admin":
    if password == "1234":
        print("Login successful")
    else:
        print("Wrong password")
else:
    print("Wrong username")
```

**nested if elif else**

```python
temperature = 32
if temperature >= 20:
    if temperature >= 30:
        print("Hot")
    elif temperature >= 25:
        print("Warm")
    else:
        print("Cool")
else:
    print("Cold")
```

**Combine conditions (and)**

```python
request_complete = True
authorised_user = True

if request_complete and authorised_user:
    print("Approve request")
```

**Combine conditions (or)**

```python
critical_incident = False
security_issue = True

if critical_incident or security_issue:
    print("Escalate issue")
```

**Combine conditions (not)**

```python
data_validated = False

if not data_validated:
    print("Run data validation")
```
