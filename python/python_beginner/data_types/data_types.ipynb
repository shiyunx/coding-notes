**String**

```python
text = "hello world!"
print (type(text))
```

**Integer**

```python
int_number = 10
print (type(int_number))
```

**Float**

```python
float_number = 10.10
print (type(float_number))
```

**Boolean**

```python
is_logged_in = True
print (type(is_logged_in))
```

**List**

```python
fruits = ["apple", "orange", "avocado"]
print (type(fruits))
```

**Tuple**

```python
colors = ("red", "blue", "green")

print(type(colors))   # tuple
print(type(colors[0]))  # str
```

**Dictionary**

```python
login = {"username": "user A", "password": "123456"}
print(type(login))
```

**Set**

```python
numbers = {1, 2, 3, 4}
print(type(numbers))
```

**None**

```python
value = None
print(type(value))
```
