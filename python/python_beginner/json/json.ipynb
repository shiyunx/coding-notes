**Python dictionary to json file**

```python
import json                              

data = {                                 
    "username": "iuser",                 
    "role": "admin"                      
}

with open("user.json", "w") as file:     
    json.dump(data, file)                # Write dictionary into json file

print("File created")                    
```

**Access json data**

**Access nested json**

```python
import json

# Create a json string
json_text = '''
{
    "username": "iuser",
    "details": {
        "email": "iuser@example.com",
        "age": 30
    }
}
'''

# Convert json string to dictionary
data = json.loads(json_text)

print(data["username"])

print(data["details"]["email"])

print(data["details"]["age"])
```

**Json list/array**

```python
import json                                      

# Create json string
json_text = '''
{
    "users": [
        {
            "username": "iuser",
            "role": "admin"
        },
        {
            "username": "juser",
            "role": "user"
        }
    ]
}
'''

# Convert jsop to Python dictionary
data = json.loads(json_text)

print(data["users"][0])

print(data["users"][0]["username"])

print(data["users"][1]["role"])
```

**Loop through json list**

```python
import json                                  

# Create a json string
json_text = '''
{
    "users": [
        {
            "username": "iuser",
            "role": "admin"
        },
        {
            "username": "juser",
            "role": "user"
        }
    ]
}
'''

# Convert json string to a python dictionary
data = json.loads(json_text)

for user in data["users"]:

    print(user["username"])

    print(user["role"])
```

**Get with default value**

```python
import json                                      

json_text = '{"username": "iuser"}'             

data = json.loads(json_text)                     # Convert json to dictionary

email = data.get("email", "No email")           

print(email)                               
```

**Check if key exists**

```python
import json                                     

json_text = '{"username": "iuser"}'              # Create json string

data = json.loads(json_text)                     # Convert json to dictionary

if "email" in data:                            
    print(data["email"])                         
else:                                            
    print("Key not found")                     
```

**Handle missing key with try/except**

```python
import json                                      

json_text = '{"username": "iuser"}'              

data = json.loads(json_text)                     # Convert json to dictionary

try:                                             # Start error-handling block
    print(data["email"])                         # Try to access missing key
except KeyError:                                 # Catch missing-key error
    print("Key not found")                       
```

**Delete a key**

```python
import json  

data = {                                         # Create python dictionary
    "username": "iuser",                      
    "password": "abc123",                      
    "is_user": True                             
}

data.pop("password", None)                     

print(data)                                      
```

**Delete key and save json file**

```python
import json                                      

with open("user.json", "r") as file:             
    user = json.load(file)                       # Convert json file to dictionary

user.pop("password", None)                      

with open("user.json", "w") as file:             
    json.dump(user, file, indent=4)              

print("Password deleted")                     
```

**Handle invalid json**

```python
import json                                     

json_text = '{"username": "iuser",}'             # Create invalid json

try:                                             # Start error-handling block
    data = json.loads(json_text)                 # Try to convert json string
    print(data)                                  

except json.JSONDecodeError:                     # Catch invalid json error
    print("Invalid JSON")                        
```
