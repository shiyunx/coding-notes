**Comparison operators**

```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
```

**Logical operators**

```python
age = 30

print(age >= 13 and age <= 65)
```

**Membership operators**

```python
fruits = ["apple", "orange"]

print("orange" in fruits)
```

**Sets**

```python
numbers = {1, 2, 3, 3, 4}

print(numbers)
```

**Add to a set**

```python
fruits = {"apple", "banana"}

fruits.add("orange")

print(fruits)
```

**Remove from a set**

```python
fruits = {"apple", "banana", "orange"}

fruits.remove("banana")
```
