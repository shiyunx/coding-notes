A variable stores a value, and the value can have different data types.

1. String `str`: text
2. Boolean `bool`: True, False
3. Integer `int`: while numbers
4. Float `float`: decimal numbers
5. List `list`: [ ]
6. Tuple `tuple`: ( )
7. Dictionary `dict`: {key: value}
8. Set `set`: {unique values}
9. None `NoneType`: no value

**Single Variables**

```python
# Store string in variable
message = "learn python"

print(message)
```

```python
# Store string in variable
name = "User"
age = 10
height = 1.15

print(name)
print(age)
print(height)
```

**Multiple Variables**

```python
# Multiple variables
first_name, last_name = "Super", "Mario"

print(first_name, last_name)
```

**String**

```python
language = "Python"
message = "Hello World"

print(language)
print(message)
```

**Integer**

```python
temperature = 30
rainfall = -5
humidity = 100

print(temperature)
print(rainfall)
print(humidity)
```

```python
x = 10
y = 5

total = x + y

print(total)
```

**Boolean**

```python
is_student = True
passed = False

print(is_student)
print(passed)
```

```python
marks = 20
is_pass = marks >= 50

print(is_pass)
```

**List**

```python
fruits = ["Apple", "Berries", "Orange"]

print(fruits)
```

```python
fruits[0] = "Mango"

print(fruits)
```

**Tuple**

```python
fruits = ("Apple", "Berries", "Orange")

print(fruits)
```

```python
print(fruits[0])
```

```python
coordinates = (10, 20)

print(coordinates)
```

**Dictionary**

```python
login = {
    "username": "Mario",
    "password": "123456",
    "status": "Logged in"
}

print(login)
```

```python
print(login["username"])
print(login["status"])
```

**Set**

```python
numbers = {1, 2, 3, 4}

print(numbers)
```

```python
numbers.add(5)

print(numbers)
```

```python
numbers = [1, 2, 2, 3, 3, 4]
unique_numbers = set(numbers)

print(unique_numbers)
```

**None**

```python
result = None

print(result)
```

```python
username = None

# Get the username
username = "123password"

print(username)
```
